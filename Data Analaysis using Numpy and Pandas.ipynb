## ** Numpy Array Operations: **

```python
import numpy as np

# Create the 1D NumPy array for Week 1 temperatures

temperature_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(f"Temperatures of week 1 in Celsius: {temperature_w1}")
```

```python
# Inspect the shape, data type, and the number of elements in the array. Inspect the shape of the temperatures_w1 array. 

print("Shape:",temperature_w1.shape)
print("Data Type:",temperature_w1.dtype)
print("Number of Elements:",temperature_w1.size)
```

```python
# Convert the temperatures_w1 from Celsius to Fahrenheit using the formula: 
# Fahrenheit = (Celsius * 9/5) + 32 

temperature_w1_f = (temperature_w1 * 9/5) + 32
print(f"Temperatures of week 1 in Fahrenheit: {temperature_w1_f}")

#Find the maximum, minimum, and mean temperatures for the week.

max_temp = temperature_w1_f.max()
print(f"Maximum Temperature: {max_temp:.2f}°F")
min_temp = temperature_w1_f.min()
print(f"Minimum Temperature: {min_temp:.2f}°F")
mean_temp = temperature_w1_f.mean()
print(f"Mean Temperature: {mean_temp:.2f}°F")
```

```python
# Temperatures for the first three days of the week.

print("First three days:",temperature_w1_f[:3])

# Temperatures for the weekend (last two days).

print("Weekend days:",temperature_w1_f[-2:])

# temperatures for the  middle three days of the week.

print("Middle three days:",temperature_w1_f[2:5])
```

```python
# Create a 2D NumPy array named temperatures where each row represents temperatures for a different week:  
# ★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)  
# ★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2) 

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Row 0: Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Row 1: Week 2
])
print(f"Temperatures of two weeks: {temperatures}")
```

```python
# Inspect the temperatures Array:

print("Shape:",temperatures.shape)
print("Data Type:",temperatures.dtype)
print("Number of Elements:",temperatures.size)

temperatures_f = (temperatures * 9/5) + 32
print(f"Temperatures of two weeks in Fahrenheit: {temperatures_f}")

max_temp_2weeks = temperatures_f.max()
print(f"Maximum Temperature: {max_temp_2weeks:.2f}°F")
min_temp_2weeks = temperatures_f.min()
print(f"Minimum Temperature: {min_temp_2weeks:.2f}°F")
mean_temp_2weeks = temperatures_f.mean()
print(f"Mean Temperature: {mean_temp_2weeks:.2f}°F")
```

```python
# Slice the temperatures Array:

print("Week 1 temperatures:",temperatures_f[0, :])
print("Weekend temperatures of Week 1:",temperatures_f[0, -2:])
print("Week 2 temperatures:",temperatures_f[1, :])
print("Weekend temperatures of Week 2:",temperatures_f[1, -2:])
```

## ** Pandas Series **

```python
# Creating Pandas Series:  

import pandas as pd
import numpy as np
# ● Create a Pandas Series named marks with the following data:  
# ➢ Marks: 95, 92, 89, 85, 80  
# ➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'  

marks = np.array([95, 92, 89, 85, 80])
indices = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
ranks = pd.Series(data = marks, index = indices)
print(ranks)
```

```python
# Indexing and Slicing:  
#●  Use the integer index position to access the mark of the 1st rank student.  

print("Mark of the first rank student is:",marks[0],"(By posiotional index)")

# ●  Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.

print("Top 3 Ranks are:", ranks.loc['Rank1':'Rank3'],"(By .loc accessor)")

# ● Use the iloc accessor to retrieve the mark of the 3rd-ranked student.

print("Mark of the 3rd rank student is:",ranks.iloc[2],"(By .iloc accessor)")

# ● Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.

above_90 = ranks > 90
ranks_above_90 = ranks[above_90]
print("Ranks with marks greater than 90 are:", ranks_above_90, sep="\n")
```

```python
# Manipulating Series:

# ● Modify the mark of the student with the 1st rank to 100.

ranks.loc['Rank1'] = 100
print(ranks.iloc[0])
print(ranks)
print(marks)

# ● Remove the entry corresponding to the last rank from the Series.

ranks = ranks.drop('Rank5')
print(ranks)
print(marks)

```

```python
# ● Compute the CGPA by dividing each mark in the Series by 10.

cgpa = ranks / 10
print("\nComputed CGPA:\n", cgpa, sep="")
```

## ** Pandas Dataframe **

```python
import pandas as pd

# Define the transaction data
data = {
    'TransactionID':[101, 102, 103, 104, 105, 106, 107, 108, 109, 110], 
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

# Create the DataFrame

transactions = pd.DataFrame(data)
print(transactions)
```

```python
# Data Exploration

# ● Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.

print("Head\n",transactions.head())
print("\nTail\n,",transactions.tail())
print(f"\nDataFrame Shape: {transactions.shape}")
print(f"\nColumn Names: {list(transactions.columns)}")
print("\nDATA TYPES & INFO")
transactions.info() # Displays non-null counts and data types for all columns
 
```

```python
# ● Display only the 'ProductCategory' and 'Amount' columns.

print(transactions[['ProductCategory', 'Amount']])
```

```python
# ● Retrieve the last 3 columns of the DataFrame using iloc or loc.

print(transactions.iloc[:,1:])
```

```python
# ● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200. 

filtered_df = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print(filtered_df)
```

```python
# ● Find the value counts for the 'ProductCategory' column.  

print(transactions['ProductCategory'].value_counts())
```

```python
# ● Find the unique values in the 'Region' column. 

print(transactions['Region'].unique())
```

```python
# ● Group by 'Region' and find the mean amount for each region.

mean_by_region = transactions.groupby('Region')['Amount'].mean()
print(mean_by_region)
```

```python
# Manipulating the DataFrame:  

# ● Modify the 'Amount' for the transaction with TransactionID 102 to 165.

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions.head(3))
```

```python
# ● Add a new column 'Discount' by calculating 10% of the 'Amount'. 

transactions['Discount'] = transactions['Amount'] * 0.10
print("Head\n",transactions.head())
```

```python
# ● Remove the row with TransactionID 109.

transactions = transactions.drop(transactions[transactions['TransactionID'] == 109].index)
print(transactions)
```

```python
# ● Delete the 'Discount' column from the DataFrame.

transactions = transactions.drop(columns=['Discount'])
print(transactions)
```
